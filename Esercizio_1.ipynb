## Overflow and underflow

```python
import numpy as np
import warnings

warnings.filterwarnings('ignore')
```

```python
def underflow(under):
    zero = type(under)(0.0)
    due  = type(under)(2.0) 
    n = 0
    while True:
        under = under/due
        if under == zero:
            break
        n += 1
    print(f"Underflow limit: 2^-{n}")

def overflow(over):
    due  = type(under)(2.0) 
    n = 0
    while True:
        over = over*due
        if np.isinf(over):
            break
        n += 1
    print(f"Overflow limit: 2^-{n}")
```

a)

```python
under   = np.float32(1.0)
over    = np.float32(1.0)

underflow(under)
overflow(over)
```

b)

```python
under   = np.float64(1.0)
over    = np.float64(1.0)

underflow(under)
overflow(over)
```
