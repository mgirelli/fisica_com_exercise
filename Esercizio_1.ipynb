## Overflow and underflow

```python
import numpy as np
import warnings

warnings.filterwarnings('ignore')
```

```python
def underflow(under):
    zero = type(under)(0.0)
    due  = type(under)(2.0) 
    n = 0
    while True:
        under = under/due
        if under == zero:
            break
        n += 1
    print(f"Underflow limit: 2^-{n}")

def overflow(over):
    due  = type(under)(2.0) 
    n = 0
    while True:
        over = over*due
        if np.isinf(over):
            break
        n += 1
    print(f"Overflow limit: 2^{n}")
```

a)

```python
under   = np.float32(1.0)
over    = np.float32(1.0)

underflow(under)
overflow(over)
```

b)

```python
under   = np.float64(1.0)
over    = np.float64(1.0)

underflow(under)
overflow(over)
```

c)

```python
def under_over_int(integer):
    due = type(integer)(2)
    uno = type(integer)(1)
    n=1
    
    while True:
        next_integer = integer * due + uno

        if next_integer <= integer:        #condizione di overflow è la variazione di segno
            break

        integer = next_integer
        n += 1
        
    print(f"Overflow limit: 2^{n} - 1")  #-1 è dato dallo zero
    print(f"Underflow limit: -2^{n} ")   
    
```

```python
val = np.int32(1)
under_over_int(val)
```

```python
val = np.int64(1)
under_over_int(val)
```

## Machine precision

```python
def precision(eps):
    due = type(eps)(2.0)
    uno = type(eps)(1.0)
    
    while True:
        
        next_eps = eps/due
        next_uno = uno + eps
        
        if next_uno == uno:
            break

        eps = next_eps
    
    print(f"Epsilon:{eps}")
```

```python
precision(np.float32(1.0))
```

## Good and bad algorithms, truncation and roundof

```python

def exp_series_i(x, tol=1e-6):
    dtype = x.dtype
    esatto = np.exp(-x, dtype=dtype)
    
    i = 0
    somma = dtype.type(0.0)
    termine = dtype.type(1.0)

    print(f"{'i':>5} | {'x':>10} | {'somma':>18} | {'errore_rel':>18}")

    while abs(termine) > tol:
        somma += termine
        i += 1
        errore_rel = abs(somma - esatto) / esatto
        print(f"{i:>5d} | {x:>10.4f} | {somma:>18.8e} | {errore_rel:>18.8e}")

        facto = np.prod(np.arange(1, i + 1, dtype=dtype))
        
        termine = (dtype.type(-1.0) ** i) * (x ** i) / facto


def exp_series_ii(x, tol=1e-6):
    dtype = x.dtype
    esatto = np.exp(-x, dtype=dtype)
    
    i = 0
    somma = dtype.type(0.0)
    termine = dtype.type(1.0)

    print(f"{'i':>5} | {'x':>10} | {'somma':>18} | {'errore_rel':>18}")

    while abs(termine) > tol:
        somma += termine
        i += 1
        errore_rel = abs(somma - esatto) / esatto
        print(f"{i:>5d} | {x:>10.4f} | {somma:>18.8e} | {errore_rel:>18.8e}")

        termine = -termine * (x / dtype.type(i))
```

```python
def exp_inverso(x, tol=1e-6):
    dtype = x.dtype
    esatto = np.exp(-x, dtype=dtype)
    
    somma = dtype.type(0.0)
    termine = dtype.type(1.0)
    i = 0
    
    while abs(termine) > tol and i < 1000:
        somma += termine
        i += 1
        termine = termine * (x / dtype.type(i))
        
    risultato = dtype.type(1.0) / somma if somma > 0 else dtype.type(0.0)
    err_rel = abs(risultato - esatto) / esatto if esatto > 0 else np.nan
    return risultato, err_rel, i
```

## Numerical analysis - Roundoff: derivative

```python
def symmetric(h):
    
    num = np.sin(1 + h) - np.sin(1 - h)
    stima = num/(2*h)
    errore = stima - np.cos(1)

    return stima, errore

def forward(h):

    num = np.sin(1 + h) - np.sin(1)
    stima = num/h
    errore = stima - np.cos(1)

    return stima, errore

def backwards(h):

    num = np.sin(1) - np.sin(1 - h)
    stima = num/h
    errore = stima - np.cos(1)

    return stima, errore
```

```python
h = np.array([0.5, 0.2, 0.1])

for i in h:
    print(f"-----{i}-----")
    print("3-point symmetric:", symmetric(i)[0], "errore:", symmetric(i)[1] )
    print("2-point “forward”:", forward(i)[0], "errore:", forward(i)[1])
    print("2-point “backwards”:", backwards(i)[0], "errore:", backwards(i)[1])
```

```python
import pandas as pd

div = np.array([10,100,1000,10000])

data = []

for d in div:
    new_h = h / d
    for h_val in new_h:
        data.append({
            "div": d,
            "h": h_val,
            "Symmetric": symmetric(h_val)[0],
            "Err Sym": symmetric(h_val)[1],
            "Forward": forward(h_val)[0],
            "Err Fwd": forward(h_val)[1],
            "Backwards": backwards(h_val)[0],
            "Err Bwd": backwards(h_val)[1],
        })

df = pd.DataFrame(data)

display(df)
```
